```python
import requests
```

```python
api_key = "51fae9cd41b30381adec85b462d9c5e4"
```

```python
url = "https://api.openweathermap.org/data/2.5/weather"
```

```python
city = input('Введите название города:')
```

```python
params = {
    'q': city,
    'appid': api_key,
    'units': 'metric'
}
```

```python
response = requests.get(url, params)
response.json()
```

```python
data = response.json()
temp = data['main']['temp']
description = data['weather'][0]['description']
print(f'Температура в городе {city}: {temp} °C')
print(f'Погода: {description}')
```
