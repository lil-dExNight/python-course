```python
import json
purchases = {}
```

```python
with open('purchase_log.txt', encoding='utf-8') as file:
    # skip {"user_id": "user_id", "category": "category"}
    next(file)
    
    for line in file:
        line = line.strip()
        if not line: continue

        rec = json.loads(line)
        purchases[rec['user_id']] = rec['category']
```

```python
import openpyxl
wb = openpyxl.load_workbook('visit_log.xlsx', read_only=True)
sheet = wb.active
```

```python
with open('funnel.csv', 'w', encoding='utf-8') as funnel:
    funnel.write('user_id,source,category\n')
    for user_id, source in sheet.iter_rows(min_row=2, values_only=True):
        user_id = str(user_id)
        if user_id in purchases:
            funnel.write(f'{user_id},{source},{purchases[user_id]}\n')
wb.close()
```
