```python
import json
purchases = {}
```

```python
with open('purchase_log.txt', encoding='utf-8') as file:
    # skip {"user_id": "user_id", "category": "category"}
    next(file)
    
    for line in file:
        line = line.strip()
        if not line: continue

        rec = json.loads(line)
        purchases[rec['user_id']] = rec['category']
```

```python
for i, item in enumerate(purchases.items()):
    if i == 2: break
    print(item[0], item[1])
```
