```python
def sum_distance(_from, _to):
    # only integers
    _from_int = _from // 1
    _to_int = _to // 1
    
    if _from_int > _to_int:
        _from_int = _to // 1
        _to_int = _from // 1

    if _from_int == _to_int:
        return _to_int

    result = 0
    while _from_int <= _to_int:
        result += _from_int
        _from_int += 1
    return result
```

```python
sum_distance(10.1, 60.000001)
```
