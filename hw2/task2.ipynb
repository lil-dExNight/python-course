```python
def trim_and_repeat(_string: str, _offset: int = 0, _repetitions: int = 1):
    # length = len(_string)
    # if length < _offset:
    #    raise -1

    trimmed_string = _string[_offset:]
    return trimmed_string * _repetitions
```

```python
trim_and_repeat("salam")
```

```python
trim_and_repeat("salam", 3)
```

```python
trim_and_repeat("salam", 3, 10)
```

```python
trim_and_repeat("salam", 5)
```

```python
trim_and_repeat("salam", 5, 10)
```
