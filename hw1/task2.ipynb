```python
men =  ['Peter', 'Alex', 'John', 'Arthur', 'Richard']
women = ['Kate', 'Liza', 'Kira', 'Emma', 'Trisha']
```

```python
if len(men) != len(women):
    print("Внимание, кто-то может остаться без пары!")
```

```python
men = sorted(men)
women = sorted(women)
```

```python
if len(men) == len(women):
    print("Идеальные пары:")
    for i in range(len(men)):
        print(f"{men[i]} и {women[i]}")
```
